```python
class Producto:
    def __init__(self, nombre="", precio=0.0, stock=0):
        self.nombre = nombre
        self.precio = precio
        self.stock = stock

    def leerdatos(self):
        print(f"\n--- Ingreso de datos del producto ---")
        self.nombre = input("Nombre del producto: ")
        self.precio = float(input("Precio: "))
        self.stock = int(input("Stock disponible: "))

    def mostrardatos(self):
        print("-" * 30)
        print(f"📦 PRODUCTO: {self.nombre}")
        print(f"💰 Precio: ${self.precio}")
        print(f"🔢 Stock: {self.stock} unidades")

# --- Creación de los 2 objetos ---
producto1 = Producto()
producto2 = Producto()

# --- Uso de la función leerdatos ---
producto1.leerdatos()
producto2.leerdatos()

# --- Uso de la función mostrardatos ---
print("\n" + "═" * 30)
print("       INVENTARIO")
print("═" * 30)
producto1.mostrardatos()
producto2.mostrardatos()
```
